```python
#difference() methodu bir setten diğer setin elemanlarını çıkarır
set1 = {"apple","banana","cherry"}
set2 = {"google","microsoft","apple"}
set3 = set1.difference(set2)# set1 den set2 deki elemanları çıkarır
print(set3)
```

```python
#difference() methodu yerine - operatörünüde kullanabiliriz
#Ama sadece setler arasında kullanılabilir
set1 = {"apple","banana","cherry"}
set2 = {"google","microsoft","apple"}
set3 = set1 - set2 # set1 den set2 deki elemanları çıkarır
print(set3)
```

```python
#difference_update() methodu ise bir seti diğer setin elemanlarını çıkararak günceller
set1 = {"apple","banana","cherry"}
set2 = {"google","microsoft","apple"}
set1.difference_update(set2)# set1 i set2 deki elemanları çıkararak günceller
print(set1)
print(set2)
```

```python
#symmetric_difference() methodu ise iki set arasındaki simetrik farkı bulur
set1 = {"apple","banana","cherry"}
set2 = {"google","microsoft","apple"}
set3 = set1.symmetric_difference(set2)# iki set arasındaki simetrik farkı bulur
print(set3)
```

```python
#symmetric_difference() methodu yerine ^ operatörünüde kullanabiliriz
#Ama sadece setler arasında kullanılabilir
set1 = {"apple", "banana" , "cherry"}
set2 = {"google", "microsoft", "apple"}
set3 = set1 ^ set2 # iki set arasındaki simetrik farkı bul
print(set3)
```

```python
#symmetric_difference_update() methodu ise bir seti diğer setle simetrik farkına göre günceller
set1 = {"apple","banana","cherry"}
set2 = {"google","microsoft","apple"}
set1.symmetric_difference_update(set2)# set1 i simetrik farka göre günceller
print(set1)
print(set2)
```

```python
#frozenset methodu ile değiştirilemeyen (immutable) setler oluşturabiliriz
#frozenset oluşturulduktan sonra eleman eklenemez veya çıkarılamaz
x = frozenset({"apple", "banana", "cherry"})
print(x)
print(type(x))
```

```python
#Frozenset metodları 
#copy()	Frozenset'in sığ bir kopyasını döndürür
#difference()	İki veya daha fazla frozenset'in farkını yeni bir frozenset olarak döndürür
#intersection()	İki veya daha fazla frozenset'in kesişimini yeni bir frozenset olarak döndürür
#isdisjoint()	İki frozenset'in ortak elemanı yoksa True döndürür
#issubset()	Başka bir frozenset bu frozenset'i içeriyorsa True döndürür
#issuperset()	Bu frozenset başka bir frozenset'i içeriyorsa True döndürür
#symmetric_difference()	İki frozenset'in simetrik farkını yeni bir frozenset olarak döndürür
#union()	İki veya daha fazla frozenset'in birleşimini yeni bir frozenset olarak döndürür
```

```python
# add() - Adds an element to the set
# clear() - Removes all elements from the set
# copy() - Returns a shallow copy of the set
# difference() (-) - Returns a set containing the difference between two or more sets
# difference_update() (-=) - Removes the items in this set that are also included in another specified set
# discard() - Removes the specified item (no error if not present)
# intersection() (&) - Returns a set that is the intersection of two or more sets
# intersection_update() (&=) - Removes the items in this set that are not present in other specified set(s)
# isdisjoint() - Returns True if two sets have no elements in common
# issubset() (<= or <) - Returns True if all items of this set are present in another set
# issuperset() (>= or >) - Returns True if all items of another set are present in this set
# pop() - Removes and returns an arbitrary element from the set
# remove() - Removes the specified element (raises KeyError if not present)
# symmetric_difference() (^) - Returns a set with the symmetric difference of two sets
# symmetric_difference_update() (^=) - Updates this set with the symmetric difference of this set and another
# union() (|) - Returns a set containing the union of sets
# update() (|=) - Update the set, adding elements from another set or iterable
```

```python
class Playlist:

    def __init__(self, name):
        self.name = name
        self.songs = []

    def add_song(self,song):
        self.songs.append(song)
        print(f"Added: {song}")
    
    def remove_song(self, song):
        if song in self.songs:
            self.songs.remove(song)
            print(f"Removed: {song}")
    
    def show_songs(self):
        print(f"Playlist '{self.name}' :")
        for song in self.songs:
            print(f"- {song}")


my_playlist = Playlist("Favorites")
my_playlist.add_song("Bohemian Rhapsody")
my_playlist.add_song("Stairway to Heaven")
my_playlist.show_songs() 
```

```python
class Car:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Drive!")

class Boat:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Sail!")

class Plane:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Fly!")

car1 = Car("Ford", "Mustang")       #Create a Car object
boat1 = Boat("Ibiza", "Touring 20") #Create a Boat object
plane1 = Plane("Boeing", "747")     #Create a Plane object

for x in (car1, boat1, plane1):
  x.move()
```

```python
class Vehicle:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Move!")

class Car(Vehicle):
  pass

class Boat(Vehicle):
  def move(self):
    print("Sail!")

class Plane(Vehicle):
  def move(self):
    print("Fly!")

car1 = Car("Ford", "Mustang")       #Create a Car object
boat1 = Boat("Ibiza", "Touring 20") #Create a Boat object
plane1 = Plane("Boeing", "747")     #Create a Plane object

for x in (car1, boat1, plane1):
  print(x.brand)
  print(x.model)
  x.move()
```

```python
class Student:
  def __init__(self, name):
    self.name = name
    self.__grade = 0

  def set_grade(self, grade):
    if 0 <= grade <= 100:
      self.__grade = grade
    else:
      print("Grade must be between 0 and 100")

  def get_grade(self):
    return self.__grade

  def get_status(self):
    if self.__grade >= 60:
      return "Passed"
    else:
      return "Failed"

student = Student("Emil")
student.set_grade(85)
print(student.get_grade())
print(student.get_status()) 
```

```python
class Calculator:
  def __init__(self):
    self.result = 0

  def __validate(self, num):
    if not isinstance(num, (int, float)):
      return False
    return True

  def add(self, num):
    if self.__validate(num):
      self.result += num
    else:
      print("Invalid number")

calc = Calculator()
calc.add(10)
calc.add(5)
print(calc.result)
# calc.__validate(5) # This would cause an error 
```

```python
class Outer:
  def __init__(self):
    self.name = "Outer Class"

  class Inner:
    def __init__(self):
      self.name = "Inner Class"

    def display(self):
      print("This is the inner class")

outer = Outer()
print(outer.name) 
```

```python
class Outer:
  def __init__(self):
    self.name = "Outer"

  class Inner:
    def __init__(self):
      self.name = "Inner"

    def display(self):
      print("Hello from inner class")

outer = Outer()
inner = outer.Inner()
inner.display() 
```

```python
class Car:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model
    self.engine = self.Engine()

  class Engine:
    def __init__(self):
      self.status = "Off"

    def start(self):
      self.status = "Running"
      print("Engine started")

    def stop(self):
      self.status = "Off"
      print("Engine stopped")

  def drive(self):
    if self.engine.status == "Running":
      print(f"Driving the {self.brand} {self.model}")
    else:
      print("Start the engine first!")

car = Car("Toyota", "Corolla")
car.drive()
car.engine.start()
car.drive() 
```
